# BTC/NQ Cross-Asset Strategy: Data and Baseline

This notebook shows the public, reproducible path from synchronized one-minute bars to baseline trade candidates. The bundled sample is intentionally small; it is used to verify the workflow rather than to prove a production trading edge.

## Scope

The target instrument is NQ. BTC is used as a cross-asset context signal. The baseline rule enters when BTC and NQ candle directions align, then exits on divergence, reversal, end-of-day safety close, or a time stop. Future information is used only to label completed trade outcomes.

```
from src.data import load_synchronized_sample
from src.features import add_market_features, feature_columns
from src.labels import generate_labeled_trades

market = load_synchronized_sample()
market.head()
```

```
features = add_market_features(market)
feature_columns(features.columns)
```

```
trades = generate_labeled_trades(features, max_holding_minutes=60)
trades[["entry_time", "entry_idx", "exit_idx", "pnl", "target"]].head()
```

## Baseline Interpretation

The baseline is deliberately simple. It is a research scaffold for evaluating whether an ML filter can reject low-quality aligned-candle signals. It does not model transaction costs, slippage, exchange outages, funding, or live execution constraints in this public sample.
